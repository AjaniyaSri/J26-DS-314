## 3. Make the station inventory and check quality

For each ArcGIS feature, we retain **both** the geometry coordinate and the `latitude`/`longitude` attributes. The latter can be rounded or inconsistent. The **distance calculations measure disagreement between sources**, not which source is correct.

All maps use GeoJSON point geometry as the *blue* ArcGIS location.

```python
rows = []
for f in ARCGIS_GEOJSON["features"]:
    p = f["properties"]
    longitude, latitude = f["geometry"]["coordinates"]  # GeoJSON order: lon, lat
    rows.append({
        "objectid": p["objectid"], "station": p["station"],
        "matched_name": NAME_ALIASES.get(p["station"], p["station"]),
        "basin": str(p.get("basin") or "").strip(),
        "tributary": str(p.get("Tributory") or "").strip(),
        "arc_lat": float(latitude), "arc_lon": float(longitude),
        "attr_lat": p.get("latitude"), "attr_lon": p.get("longitude"),
        "alert_level": p.get("Alert_Level"),
        "minor_flood_level": p.get("Minor_Flood_Level"),
        "major_flood_level": p.get("Major_Flood_Level"),
        "unit": p.get("Unit"),
        "elevation_m_msl": p.get("Elivation_m_MSL"),
    })

stations = pd.DataFrame(rows)
# Match by station name, not by ID or row order: these differ across the two sources.
comparison = stations.merge(
    annual, left_on="matched_name", right_on="station", how="left",
    suffixes=("", "_pdf"), validate="one_to_one", indicator=True
)

def geodesic_m(lon1, lat1, lon2, lat2):
    return GEOD.inv(lon1, lat1, lon2, lat2)[2]

comparison["geom_vs_attr_m"] = comparison.apply(
    lambda r: geodesic_m(r.arc_lon, r.arc_lat, r.attr_lon, r.attr_lat)
    if pd.notna(r.attr_lat) and pd.notna(r.attr_lon) else float("nan"), axis=1
)
comparison["arc_vs_pdf_m"] = comparison.apply(
    lambda r: geodesic_m(r.arc_lon, r.arc_lat, r.pdf_lon, r.pdf_lat)
    if pd.notna(r.pdf_lat) else float("nan"), axis=1
)
comparison["arc_vs_kand_m"] = comparison.apply(
    lambda r: geodesic_m(r.arc_lon, r.arc_lat, r.kand_lon, r.kand_lat)
    if pd.notna(r.kand_lat) else float("nan"), axis=1
)
comparison["review_flag"] = comparison.apply(
    lambda r: "NO ANNUAL MATCH" if pd.isna(r.pdf_lat)
    else ("GEOMETRY/ATTRIBUTE CONFLICT" if r.geom_vs_attr_m > 25
          else ("SOURCE DISAGREEMENT >= 100 m" if r.arc_vs_pdf_m >= 100 else "SOURCE COMPARISON OK (<100 m)")),
    axis=1
)

print("ArcGIS feature points:", len(stations))
print("Matched against 40 annual-report records:", (comparison._merge == "both").sum())
print("Additional ArcGIS stations:", comparison.loc[comparison._merge.eq("left_only"), "station"].tolist())
print("ArcGIS geometry vs attribute conflicts >25 m:",
      comparison.loc[comparison.geom_vs_attr_m > 25, "station"].tolist())
print("\nTOP DISTANCES: ArcGIS geometry vs published WGS84")
display(comparison.sort_values("arc_vs_pdf_m", ascending=False)[
    ["station", "basin", "arc_vs_pdf_m", "arc_vs_kand_m", "geom_vs_attr_m", "review_flag"]
].head(12).round(1))
```

## 4. National interactive map — all ArcGIS station geometries

Map layers (toggle in top-right control):

- **Blue pins:** ArcGIS **GeoJSON point geometry**, enabled by default (42 stations).
- **Red circles:** PDF published WGS84 positions (40 stations), optional.
- **Orange circles:** PDF Kandawala positions after conversion to WGS84 (40 stations), optional.

Zoom toward a station and turn layers on/off. Clicking an ArcGIS marker shows the original station name, basin, tributary, coordinates, and flood level thresholds with their **original units**. These thresholds are included as *source attributes*, not interpreted or recalibrated. Basemap layers come from OSM/Esri and load online.

```python
ESRI_URL = (
    "https://server.arcgisonline.com/ArcGIS/rest/services/"
    "World_Imagery/MapServer/tile/{z}/{y}/{x}"
)
ESRI_ATTRIBUTION = "Tiles &copy; Esri, Maxar, Earthstar Geographics, and GIS User Community"

def make_base_map(center, zoom=8, satellite_first=False, height=550):
    m = folium.Map(location=center, zoom_start=zoom, tiles=None,
                   control_scale=True, height=height)
    folium.TileLayer("OpenStreetMap", name="OpenStreetMap", show=not satellite_first).add_to(m)
    folium.TileLayer(tiles=ESRI_URL, attr=ESRI_ATTRIBUTION,
                     name="Esri Satellite", show=satellite_first, max_zoom=19).add_to(m)
    return m

def popup_arcgis(r):
    st, basin, tributary = [html.escape(str(x)) for x in
                           (r.station, r.basin, r.tributary)]
    def field(label, value):
        return f"<b>{label}:</b> {html.escape(str(value)) if pd.notna(value) else 'Not provided'}<br>"
    return folium.Popup(
        f"<b>{st}</b><br>{basin} / {tributary}<br>"
        f"<b>ArcGIS geometry:</b> {r.arc_lat:.8f}, {r.arc_lon:.8f}<br>"
        + field("Alert", str(r.alert_level) + ' ' + str(r.unit) if pd.notna(r.alert_level) else None)
        + field("Minor flood", str(r.minor_flood_level) + ' ' + str(r.unit) if pd.notna(r.minor_flood_level) else None)
        + field("Major flood", str(r.major_flood_level) + ' ' + str(r.unit) if pd.notna(r.major_flood_level) else None)
        + f"<b>Verification:</b> Not independently field-verified",
        max_width=350,
    )

national = make_base_map([7.45,80.75],zoom=8)
blue = folium.FeatureGroup(name="ArcGIS point geometry — BLUE (42)", show=True)
blue_cluster = plugins.MarkerCluster(name="ArcGIS stations", show=True).add_to(blue)
for r in comparison.itertuples(index=False):
    folium.Marker([r.arc_lat, r.arc_lon],
                  tooltip=f"{r.station} — ArcGIS geometry",
                  popup=popup_arcgis(r),
                  icon=folium.Icon(color="blue", icon="tint", prefix="fa")
                 ).add_to(blue_cluster)
blue.add_to(national)

red = folium.FeatureGroup(name="Published PDF WGS84 — RED (40)", show=False)
orange = folium.FeatureGroup(name="PDF Kandawala transformed — ORANGE (40)", show=False)
for r in annual.itertuples(index=False):
    folium.CircleMarker([r.pdf_lat,r.pdf_lon],radius=5,color="#ab1f23",
        fill=True,fill_color="#db3d3d",fill_opacity=0.9,
        tooltip=f"{r.station} — PDF WGS84",
        popup=f"{html.escape(r.station)} — PDF WGS84: {r.pdf_lat:.8f}, {r.pdf_lon:.8f}").add_to(red)
    folium.CircleMarker([r.kand_lat,r.kand_lon],radius=4,color="#8c5200",
        fill=True,fill_color="#ff9900",fill_opacity=0.9,
        tooltip=f"{r.station} — Kandawala converted",
        popup=f"{html.escape(r.station)} — Kandawala-derived (NOT surveyed)").add_to(orange)
red.add_to(national)
orange.add_to(national)
folium.LayerControl(collapsed=False).add_to(national)
plugins.Fullscreen().add_to(national)
plugins.MousePosition(position="bottomleft",num_digits=7,
                      separator=" | ",prefix="Cursor lat/lon:").add_to(national)
display(national)
```
